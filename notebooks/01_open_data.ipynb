```python
import geopandas as gpd
import pandas as pd
from shapely.geometry import Point, Polygon
import matplotlib.pyplot as plt

print("geopandas", gpd.__version__)
```

## Step 2 — The research question 研究問題

**Question 問題**: *Where are the greener public spaces?* 哪裡是較綠意盎然的公共空間？

**Required variables 所需變數**: tree locations (points), public space boundaries (polygons) 樹木位置（點）、公共空間邊界（面）

**Potential dataset 可能資料集**: street tree inventory, park/plaza polygons 行道樹清冊、公園／廣場面資料

**Data source (real world) 資料來源（真實世界）**: city open data portal, OpenStreetMap 城市開放資料平台、OpenStreetMap

## Step 3 — Load a small reproducible sample dataset 載入小型可重現的範例資料集

In a real project you would download this from OSM or a government portal (Step 5 below shows how). Here we build a small, self-contained sample by hand — inspired by a generic Taiwanese urban public-space context — so the notebook runs the same way for everyone, with no network dependency.
在真實專案中，你會從 OSM 或政府平台下載這類資料（見下方步驟 5）。這裡我們手動建立一個小型、自足的範例資料集——以一般台灣都市公共空間情境為靈感——讓筆記本對每個人都能得到一致結果，且不依賴網路連線。

```python
# Sample street trees (points) — lon/lat near a generic Taipei public space
# 範例行道樹（點）— 位於一般化台北公共空間附近的經緯度
tree_coords = [
    (121.5320, 25.0335), (121.5325, 25.0338), (121.5330, 25.0330),
    (121.5340, 25.0345), (121.5345, 25.0348), (121.5300, 25.0320),
    (121.5305, 25.0322), (121.5360, 25.0360),
]
trees = gpd.GeoDataFrame(
    {"tree_id": range(1, len(tree_coords) + 1), "species": "unknown"},
    geometry=[Point(xy) for xy in tree_coords],
    crs="EPSG:4326",  # WGS84 — standard lon/lat CRS for open geodata
)

# Sample public space polygons (park, plaza)
# 範例公共空間面（公園、廣場）
public_spaces = gpd.GeoDataFrame(
    {"name": ["Riverside Park", "Central Plaza", "Corner Lot Square"]},
    geometry=[
        Polygon([(121.5315, 25.0328), (121.5335, 25.0328), (121.5335, 25.0350), (121.5315, 25.0350)]),
        Polygon([(121.5295, 25.0315), (121.5310, 25.0315), (121.5310, 25.0328), (121.5295, 25.0328)]),
        Polygon([(121.5352, 25.0352), (121.5365, 25.0352), (121.5365, 25.0365), (121.5352, 25.0365)]),
    ],
    crs="EPSG:4326",
)

trees.head()
```

**Expected output 預期輸出**: a table with columns `tree_id`, `species`, `geometry` — 8 rows, each geometry a `POINT (...)`.
一張含 `tree_id`、`species`、`geometry` 欄位的表格，共 8 列，每個幾何為 `POINT (...)`。

```python
fig, ax = plt.subplots(figsize=(6, 6))
public_spaces.plot(ax=ax, color="#c7e9c0", edgecolor="black")
trees.plot(ax=ax, color="#238b45", markersize=30)
ax.set_title("Sample public spaces and trees 範例公共空間與樹木")
plt.show()
```
